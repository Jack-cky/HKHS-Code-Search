```python
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
src = "../checkpoints/bge-small-en-v1.5-ft-hs/v0.3"
dst = "../checkpoints/bge-small-en-v1.5-ft-hs/v0.3-onnx"

model = SentenceTransformer(src, backend="onnx")
model.save_pretrained(dst)
```

```python
ref = SentenceTransformer(src).encode(["iPhone Duo"])
out = SentenceTransformer(dst, backend="onnx").encode(["iPhone Duo"])
assert np.allclose(ref, out, atol=1e-4)
```
